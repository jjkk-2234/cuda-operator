```python
import torch

# torch version
def softmax(x, dim=-1):
    x_max = x.max(dim=dim, keepdim=True)[0]

    safe_x = x - x_max

    exp_x = torch.exp(safe_x)

    sum_exp = exp_x.sum(dim=dim, keepdim=True)

    output = exp_x / sum_exp

    return output

if __name__ == '__main__':
    x1 = torch.tensor([1.0, 2.0, 3.0])
    p1 = softmax(x1)
    print(torch.softmax(x1, dim=-1))
```
